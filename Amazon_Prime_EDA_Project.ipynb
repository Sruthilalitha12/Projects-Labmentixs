## 3. ***Data Wrangling***

### Data Wrangling Code

```python
# Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import ast # Import ast here as it's used later

# Load Dataset
from google.colab import drive
drive.mount('/content/drive')

# Load the datasets
credits_df = pd.read_csv('/content/drive/MyDrive/Sruthi Lalitha Projects/Amazon(25-05-2025)/credits.csv.zip')
titles_df = pd.read_csv('/content/drive/MyDrive/Sruthi Lalitha Projects/Amazon(25-05-2025)/titles.csv.zip')

#merging the datasets
merged_df = pd.merge(credits_df, titles_df, on='id', how="left")

# Before handling missing values, inspect the columns
print("Columns in credits_df:", credits_df.columns)
print("Columns in titles_df:", titles_df.columns)
print("Columns in merged_df:", merged_df.columns)

#remove duplicate values
merged_df = merged_df.drop_duplicates()

# handling missing values
#filling missing charactor with 'unknown' for actor
if 'name' in merged_df.columns:
    merged_df['name'] = merged_df['name'].fillna('Unknown')
elif 'actor' in merged_df.columns: # Keep the original line if 'actor' somehow exists
     merged_df['actor'] = merged_df['actor'].fillna('Unknown')
else:
    print("Could not find 'actor' or 'name' column in merged_df. Please check the column names.")


#filling missing director with 'unknown'
if 'character' in merged_df.columns:
    merged_df['character'] = merged_df['character'].fillna('Unknown')

if 'director' in merged_df.columns:
    merged_df['director'] = merged_df['director'].fillna('Unknown')
else:
    print("Could not find 'director' column in merged_df. Please check the column names.")


#filling missing description with 'No description available'
# Assign the result of fillna back to the column
if 'description' in merged_df.columns:
    merged_df['description'] = merged_df['description'].fillna('No description available')
else:
    print("Could not find 'description' column in merged_df. Please check the column names.")

#filling missing age verification with 'Unrated', column name corrected based on previous info() output
# Assign the result of fillna back to the column
if 'age_certification' in merged_df.columns:
    merged_df['age_certification'] = merged_df['age_certification'].fillna('Unrated')
else:
    print("Could not find 'age_certification' column in merged_df. Please check the column names.")


#fill missing numerical values with appropriate substitutes
# Assign the result of fillna back to the column
if 'seasons' in merged_df.columns:
    merged_df['seasons'] = merged_df['seasons'].fillna(0)
else:
     print("Could not find 'seasons' column in merged_df. Please check the column names.")

if 'imdb_score' in merged_df.columns:
    merged_df['imdb_score'] = merged_df['imdb_score'].fillna(merged_df['imdb_score'].mean())
else:
    print("Could not find 'imdb_score' column in merged_df. Please check the column names.")

if 'imdb_votes' in merged_df.columns:
    merged_df['imdb_votes'] = merged_df['imdb_votes'].fillna(merged_df['imdb_votes'].mean())
else:
     print("Could not find 'imdb_votes' column in merged_df. Please check the column names.")

if 'tmdb_popularity' in merged_df.columns:
    merged_df['tmdb_popularity'] = merged_df['tmdb_popularity'].fillna(merged_df['tmdb_popularity'].mean())
else:
     print("Could not find 'tmdb_popularity' column in merged_df. Please check the column names.")

if 'tmdb_score' in merged_df.columns:
    merged_df['tmdb_score'] = merged_df['tmdb_score'].fillna(merged_df['tmdb_score'].mean())
else:
    print("Could not find 'tmdb_score' column in merged_df. Please check the column names.")

#standardize data formats
#convert genres and production countries to lists if they are in string format

def convert_to_list(x):
  if isinstance(x,str):
    try:
        return ast.literal_eval(x)
    except (ValueError, SyntaxError): # Handle cases where the string isn't a valid list representation
        return []
  else:
    return []

if 'genres' in merged_df.columns:
    merged_df['genres']=merged_df['genres'].apply(convert_to_list)
else:
    print("Could not find 'genres' column in merged_df. Please check the column names.")

if 'production_countries' in merged_df.columns:
    merged_df['production_countries']=merged_df['production_countries'].apply(convert_to_list)
else:
    print("Could not find 'production_countries' column in merged_df. Please check the column names.")

# convert data types
if 'release_year' in merged_df.columns:
    merged_df['release_year']=merged_df['release_year'].astype(int)
else:
    print("Could not find 'release_year' column in merged_df. Please check the column names.")

if 'runtime' in merged_df.columns:
    merged_df['runtime']=merged_df['runtime'].astype(int)
else:
    print("Could not find 'runtime' column in merged_df. Please check the column names.")

if 'seasons' in merged_df.columns:
    merged_df['seasons']=merged_df['seasons'].astype(int)
else:
     print("Could not find 'seasons' column in merged_df. Please check the column names.")

# merged_df['imdb_votes']=merged_df['imdb_votes'].astype(int) # This column might contain float values after fillna, converting to int might cause issues if there are NaNs or non-integer floats. Let's keep it float for now unless specifically needed as int.
if 'tmdb_popularity' in merged_df.columns:
    merged_df['tmdb_popularity']=merged_df['tmdb_popularity'].astype(float)
else:
     print("Could not find 'tmdb_popularity' column in merged_df. Please check the column names.")


#confirm changes
merged_df.info() # Only show info() for clarity
merged_df.head()

```

### What all manipulations have you done and insights you found?

*  Have removed 168 duplicate rows and reduced the dataset to 124,179 rows.
*  Handled missing values  
-Filled character with 'Unknown'for actors.
-Replaced missing description with 'No description available'.
-Set age_certification as 'Unrated' for missing values.
-Set seasons to 0 for movies.
-Filled missing imdb_score and tmdb_score with their median values.
-Set missing imdb_votes and tmdb_popularity to 0.
-Converted genres and production_countries to list format for better analysis.

###**Hypothetical Testing-1**
####**Statement-1**

a. State the research hypothesis as a null hypothesis and alternative hypothesis.

Null Hypothesis(H<sub>0</sub>): The average IMDB score for Movies and TV shows are the same.

Alternate Hypothesis(H<sub>a</sub>): The average IMDB score for Movies and TV shows are different.

b. Performing an appropriate statistical test.

```python
#Perform statistical test to obtain P-value
import scipy.stats as stats

#Split the data into Movies and TV Shows
movies_scores = merged_df[merged_df['type'] == 'MOVIE']['imdb_score']
tv_shows_scores = merged_df[merged_df['type'] == 'TV SHOW']['imdb_score']

#Perform independent t-test
t_statistic, p_value = stats.ttest_ind(movies_scores, tv_shows_scores, equal_var=False)

#Print results
print("Hypothesis-1: IMDB Scores of Movies vs TV Shows")
print(f"T-statistic: {t_statistic:.4f}, P-value:{p_value:.4f}")
if p_value < 0.05:
    print("Reject the null hypothesis. There is a significant difference in IMDB scores between Movies and TV Shows.")
else:
    print("Fail to reject the null hypothesis. There is no significant difference in IMDB scores between Movies and TV Shows.")
```

###**Hypothetical Testing-2**
####**Statement-2**

a. State the research hypothesis as a null hypothesis and alternative hypothesis.

Null Hypothesis(H<sub>0</sub>): There is no correlation between runtime and IMDb scores.

Alternate Hypothesis(H<sub>a</sub>): There is a positive correlation between runtime and IMDb scores.

b. Performing an appropriate statistical test.

```python
#Perform statistical test to obtain P-value
from scipy.stats import pearsonr

#Calculate Pearson correlation coefficient and p-value
correlation, p_value = pearsonr(merged_df['runtime'], merged_df['imdb_score'])

#Print results
print("Hypothesis-2: Correlation between Runtime and IMDB Scores")
print(f"Correlation Coefficient: {correlation:.4f}, P-value: {p_value:.4f}")
if p_value < 0.05:
    print("Reject the null hypothesis. There is a positive correlation between runtime and IMDB scores.")
else:
    print("Fail to reject the null hypothesis. There is no significant correlation between runtime and IMDB scores.")
```
