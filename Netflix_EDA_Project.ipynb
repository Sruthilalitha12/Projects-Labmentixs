# **Project Summary -**

Netflix has transformed the entertainment industry by redefining how audiences consume content. This project investigates how Netflix's catalog has evolved over time, focusing on trends in movie and TV show availability. Using **Exploratory Data Analysis (EDA)**, this study leverages a dataset sourced from **Flixable**, a third-party Netflix search engine, to understand how Netflix's strategy has shifted between 2010 and 2019.

**Understanding the Content Shift**

The dataset indicates that the number of TV shows on Netflix has increased dramatically over the years, while the number of movies has steadily declined. This shift suggests a strategic pivot towards serialized content, likely due to changing audience preferences and Netflix's expansion into original programming. TV shows often encourage higher user retention since viewers remain engaged across multiple episodes, unlike standalone movies. By analyzing trends in the dataset, we aim to uncover whether Netflix's focus on TV shows is part of a deliberate strategy or a natural industry evolution.

**Regional Availability and Content Preferences**

One of the core aspects of the analysis is evaluating **content distribution across different countries**. Netflix operates in numerous regions, and its catalog varies depending on licensing agreements, regional preferences, and government regulations. By breaking down content availability by country, we aim to understand how Netflix tailors its offerings to meet local demand. Some regions may have a higher concentration of TV shows due to local preferences, while others may prioritize movies based on viewer trends. Examining the geographical distribution of content can shed light on Netflix’s business strategy and content localization efforts.

**Impact of External Ratings and User Preferences**

To enhance the analysis, we integrate data from external sources such as IMDb ratings and Rotten Tomatoes scores to evaluate how content quality correlates with Netflix's strategic decisions. By comparing ratings of Netflix’s movies and TV shows, we can determine whether user preferences align with Netflix’s increasing focus on TV series. If TV shows receive higher ratings and engagement metrics than movies, this could validate Netflix’s push towards serialized content. Conversely, if highly rated movies are being phased out, it might indicate a gap in Netflix’s content strategy that could be explored further.

**Analytical Approaches and Insights**

Several **data-driven methods** are used to uncover patterns in content trends:
- **Time-series analysis** tracks changes in the number of movies and TV shows on Netflix over the years.
- **Country-wise segmentation** helps understand regional preferences and content distribution strategies.
- **Genre distribution analysis** evaluates how the types of movies and TV shows have evolved, identifying dominant genres.
- **Correlation studies** with external ratings assess the impact of content changes on viewer satisfaction.
By leveraging structured data analysis techniques, this project provides valuable insights into Netflix's evolving content strategy and its implications for the streaming industry. Understanding these trends can help predict future shifts in content distribution and guide streaming platforms in optimizing their offerings.

```python
# converting date_added column to date time and extracting the day, month and year from the column
df['date_added'] = pd.to_datetime(df['date_added'], format='%B %d %Y', errors='coerce')

#extracting day, month and year
df['added_day'] = df['date_added'].dt.day

df['added_month'] = df['date_added'].dt.month

df['added_year'] = df['date_added'].dt.year
```

# **5. Hypothetical Testing**

**Statement-1: Movies are added to Netflix more frequently than TV Shows.**

Here we test this using a two-sample t-test comparing the number of movies and TV shows added per year.

1. Group the data by year_added and type (Movie/TV Show).

2. Count the number of each content type added per year.

3. Perform a two-sample t-test to see if there's a significant difference.

```python
import pandas as pd
import scipy.stats as stats

df['date_added'] = pd.to_datetime(df['date_added'], errors='coerce')
df['year_added'] = df['date_added'].dt.year

df_clean = df.dropna(subset=['year_added', 'type'])

added_per_year = df_clean.groupby(['year_added', 'type']).size().unstack(fill_value=0)

print(added_per_year.head())

if 'Movie' in added_per_year.columns and 'TV Show' in added_per_year.columns:
    movie_counts = added_per_year['Movie']
    tv_counts = added_per_year['TV Show']

    t_stat, p_value = stats.ttest_ind(movie_counts, tv_counts, equal_var=False)

    print("\nT-statistic:", t_stat)
    print("P-value:", p_value)

    alpha = 0.05
    if p_value < alpha:
        print("✅ Reject the null hypothesis: Movies are added more frequently than TV Shows.")
    else:
        print("❌ Fail to reject the null hypothesis: No significant difference in addition frequency.")
else:
    print("One of the types ('Movie' or 'TV Show') is missing in the data.")
```

**Interpretation:**

- T-statistic and P-value indicate how different the groups are.
- Final print statement interprets the statistical result.

**Statement-2: There is a significant difference in content ratings between Movies and TV Shows on Netflix.**


Here we will use the Chi-Square Test of Independence because:

1. We're comparing categorical variables: rating and type.
2. It checks whether there's an association between the two.

```python
import pandas as pd
from scipy.stats import chi2_contingency


df_clean = df.dropna(subset=['rating', 'type'])

contingency_table = pd.crosstab(df_clean['type'], df_clean['rating'])

print("Contingency Table:\n", contingency_table)

chi2, p_value, dof, expected = chi2_contingency(contingency_table)

print("\nChi-Square Statistic:", chi2)
print("Degrees of Freedom:", dof)
print("P-value:", p_value)

alpha = 0.05
if p_value < alpha:
    print("✅ Reject the null hypothesis: Content ratings differ significantly between Movies and TV Shows.")
else:
    print("❌ Fail to reject the null hypothesis: No significant difference in content ratings between Movies and TV Shows.")
```

**Interpretation:**

- Null Hypothesis (H₀): Rating distribution is independent of type (Movie/TV Show).

- Alternative Hypothesis (H₁): Rating distribution depends on the type of content

**Statement-3: More than 25% of Netflix content is from the United States.**

Here we test whether the proportion of content from the United States is significantly higher than the average proportion from other countries using a One-Sample Proportion Z-Test.

```python
import numpy as np
from statsmodels.stats.proportion import proportion_confint

df['country'] = df['country'].fillna('Unknown').astype(str)
country_split = df['country'].str.split(', ').explode()

total = len(country_split)

us_count = country_split[country_split == 'United States'].count()

us_proportion = us_count / total

lower, upper = proportion_confint(us_count, total, alpha=0.05, method='normal')

print(f"Proportion of US content: {us_proportion:.4f}")
print(f"95% Confidence Interval: ({lower:.4f}, {upper:.4f})")

if lower > 0.25:
    print("✅ Reject the null hypothesis: More than 25% of Netflix content is from the US.")
else:
    print("❌ Fail to reject the null hypothesis: US content does not significantly exceed 25%.")
```

**Interpretation:**

- Null Hypothesis (H₀): 25% or less of Netflix content is from the US.
- Alternative Hypothesis (H₁): More than 25% of Netflix content is from the US.
